### Building Markov transition matrix (1 mark)

Importing Libraries

```python
import pandas as pd
import math
```

Declaring sequence and states

```python
sequence = "AGCTTAGCTAGGCTAACGTTAGCACCCGGTTACGGTTACGTTGCATGCATGTGTGCACTCTGA"

states = ['A', 'C', 'G', 'T']

counts = pd.DataFrame(0, index=states, columns=states)
```


Iterate through the sequence to count dinucleotide transitions

```python
for i in range(len(sequence) - 1):
    prev_state = sequence[i]
    next_state = sequence[i+1]
    if prev_state in states and next_state in states:
        counts.loc[prev_state, next_state] += 1
```

Compute transition probabilities by normalizing counts row-wise

```python
total_transitions = counts.sum(axis=1)
transition_matrix = counts.div(total_transitions, axis=0)
```

Print the resulting matrices

```python
print("Transition Count Matrix:")
print(counts)

print("\nTransition Probability Matrix:")
print(transition_matrix)
```
