### Multi-line sequence fasta to single-line fasta file (1 mark)

Script that takes input from multilint_inpute.fasta and output in single_line_input.fasta

```python
with open('multiline_input.fasta', 'r') as infile, open('single_line_input.fasta', 'w') as outfile:
    for line in infile:
        if line.startswith('>'):
            if outfile.tell() == 0:
                outfile.write(line.strip() + '\n') # handle the first line separately
            else :
                outfile.write('\n' + line.strip() + '\n') # new sequence in a new line
        else:
            outfile.write(line.strip()) # strip newlines and write in the same line
```
